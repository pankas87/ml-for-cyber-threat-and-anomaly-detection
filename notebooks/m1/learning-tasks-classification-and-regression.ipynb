```python
from sklearn import linear_model

X = [[1], [1], [1.5], [2.5], [2.5], [2.5], [2.5], [3.5], [3.5], [4.5], [4.5]] # Mary's ratings
Y = [5, 4, 4, 5, 3, 1.5, 1, 5, 4, 5, 4] # John's ratings 

reg = linear_model.LinearRegression()
reg.fit(X, Y)

print("Coefficients:", reg.coef_)
print("Intercept:", reg.intercept_)

reg_pred = reg.predict(X) # Save predictions in reg_pred

print("Predictions on training data:", reg_pred)
print("Prediction on Gravity [3]:", reg.predict([[3]]))
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, Y, color='black')
plt.plot(X, reg_pred, color='blue', linewidth=3)
```

```python
from sklearn.metrics import mean_squared_error

print("Mean Squared Error on training data:", mean_squared_error(Y, reg_pred))
print("Mean Squared Error on movie Gravity [3]:", mean_squared_error(reg.predict([[3]]), [3]))
```
